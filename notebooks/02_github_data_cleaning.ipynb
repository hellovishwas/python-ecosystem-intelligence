## 5. License Cleaning and Categorization

### License Standardization

```python
github_df["license"].value_counts()
```

### License Category Assignment

```python
def clean_licenses(df):
    df["license"] = df["license"].astype(str).str.strip()
    mapping = {
        "Unknown": "Unspecified",
        "NOASSERTION": "Unspecified",
        "BSD-3-Clause-Clear": "BSD-3-Clause",
    }
    df["license_cleaned"] = df["license"].replace(mapping)
    return df


def get_license_category(lic):
    permissive = [
        "MIT",
        "Apache-2.0",
        "BSD-3-Clause",
        "BSD-2-Clause",
        "ISC",
        "MIT-0",
        "Unlicense",
        "CC0-1.0",
        "WTFPL",
        "0BSD",
        "BSL-1.0",
        "Zlib",
    ]
    copyleft = ["GPL-3.0", "GPL-2.0", "AGPL-3.0", "LGPL-3.0", "LGPL-2.1", "EUPL-1.2"]

    if lic in ["Unknown", "NOASSERTION", "Unspecified"]:
        return "Unknown / Risk"
    elif lic in permissive:
        return "Permissive (Safe)"
    elif lic in copyleft:
        return "Copyleft (Risky for Commercial)"
    else:
        return "Other"


github_df = clean_licenses(github_df)
github_df["license_category"] = github_df["license_cleaned"].apply(get_license_category)
```

## 9. Outlier Detection

### IQR-Based Outlier Detection

The Interquartile Range (IQR) method is used to identify potential outliers in repository-level metrics such as stars, forks, repository size, and open issues.

Values below `Q1 − 1.5 × IQR` or above `Q3 + 1.5 × IQR` are flagged as potential outliers. These observations are inspected rather than automatically removed, as highly popular repositories can naturally have unusually large values.


```python
categories = ["stars", "forks", "repo_size", "open_issues"]
```

```python
Q1=github_df[categories].quantile(0.25)
Q3=github_df[categories].quantile(0.75)
print("Q1:\n",Q1)
print("Q3:\n",Q3)
```

```python
IQR=Q3-Q1
print("IQR:\n",IQR)
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
print("Lower Limit:\n",lower_limit)
print("Upper Limit:\n",upper_limit)
```

```python
for category in categories:
    outliers=github_df[(github_df[category]<lower_limit[category]) | ( github_df[category]>upper_limit[category])]
    print(f"Outliers in {category}: {len(outliers)}")
```
