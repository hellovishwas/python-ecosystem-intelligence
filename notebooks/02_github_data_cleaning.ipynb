## Missing Value Treatment


```python
github_df[["description","license"]]=github_df[["description","license"]].fillna("Unknown")
```

```python
github_df["language"]=github_df["language"].fillna("Python")
```

## Missing Value Verification

```python
github_df.isnull().sum()
```

## Data Type Conversion

```python
github_df["created_at"]=pd.to_datetime(github_df["created_at"])
```

```python
github_df["updated_at"]=pd.to_datetime(github_df["updated_at"])
```

```python
github_df.dtypes
```

## Numerical Data Validation

```python
columns = github_df.select_dtypes(include=["number"]).columns.tolist()
for column in columns:
    if column=="created_year":
        invalid_year = (github_df[column] > 2026) |( github_df[column] < 2020)
        print(f"Invalid {column}:",invalid_year.sum())
    else:
        invalid_=github_df[column]<0
        print(f"Invalid {column}:",invalid_.sum())
```

## Date Consistency


```python
inconsistent_date=github_df["created_at"]>github_df["updated_at"]
print(inconsistent_date.sum())
```

## Categorical Data Consistency


```python
categorical_col=["language","license","visibility","default_branch"]
for col in categorical_col:
    unique=github_df[col].nunique()
    print(f"Unique Values in {col}:",unique)
    print(github_df[col].value_counts())
    print()
```

## Outlier Detection

```python
categories = ["stars", "forks", "watchers", "repo_size", "open_issues"]
```

```python
Q1=github_df[categories].quantile(0.25)
Q3=github_df[categories].quantile(0.75)
print("Q1:\n",Q1)
print("Q3:\n",Q3)
```

```python
IQR=Q3-Q1
print("IQR:\n",IQR)
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
print("Lower Limit:\n",lower_limit)
print("Upper Limit:\n",upper_limit)
```

```python
for category in categories:
    outliers=github_df[(github_df[category]<lower_limit[category]) | ( github_df[category]>upper_limit[category])]
    print(f"Outliers in {category}: {len(outliers)}")
```

## Relationship Consistency

```python
forks_vs_stars=github_df["forks"]>github_df["stars"]
print(forks_vs_stars.sum())
```

```python
watchers_vs_stars=github_df["watchers"]!=github_df["stars"]
print(watchers_vs_stars.sum())
```
