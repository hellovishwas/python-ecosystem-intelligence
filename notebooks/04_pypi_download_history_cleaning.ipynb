```python
data = r"..\data\raw\pypi_download_history.csv"
df=pd.read_csv(data)
```

## 3. Date Conversion

```python
df["date"]=pd.to_datetime(df["date"])
```

```python
df["date"].dtypes
```

## 4. Missing Values and Duplicate Check

```python
df.isnull().sum()
```

```python
print("No. of dulicated rows:",df.duplicated().sum())
```

## 5. Package Name Consistency

```python
df["package_name"].value_counts()
```

```python
df["package_name"]=df["package_name"].astype(str).str.strip()
```

## 6. Download Data Inspection

```python
df["downloads"].describe().reset_index()
```

## 7. Dataset Ordering

```python
df=df.sort_values("date").reset_index(drop=True)
```

## 8. Final Dataset Export

```python
df.to_csv(r"..\data\processed\pypi_download_history_cleaned.csv",index=False)
```
